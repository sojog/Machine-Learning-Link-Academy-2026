```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = np.arange(1, 101)
y = 2 * x + 1

plt.scatter(x, y)
```

```python
x_train = x[:80]
y_train = y[:80]


x_test = x[80:]
y_test = y[80:]

plt.scatter(x_train, y_train)
plt.scatter(x_test, y_test)
```
