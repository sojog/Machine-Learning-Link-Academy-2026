```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# (mp, pret)


apartamente = np.array([(37, 62_000), (45, 57_000), (50, 80_000), (72, 155_000), (80, 170_000), (90, 110_000), (250, 400_000), (500, 1_200_000)])
apartamente
```

```python
# X
suprafete = apartamente[ :, 0 ]

# Y
preturi = apartamente[ :, 1 ]
```

```python
# Vizualizare
plt.scatter(suprafete, preturi)
plt.xlabel("Suprafete")
plt.ylabel("Preturi")
```

# Varianta cu sklearn

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model
```

```python
suprafete_2d = suprafete.reshape(-1, 1)
suprafete_2d
```

```python
model.fit(suprafete_2d, preturi)
```

```python
# Vizualizare
plt.scatter(suprafete, preturi)
plt.xlabel("Suprafete")
plt.ylabel("Preturi")
plt.plot(suprafete, model.predict(suprafete_2d))
```

```python
# Vizualizare
plt.scatter(suprafete, preturi)
plt.xlabel("Suprafete")
plt.ylabel("Preturi")
plt.plot(suprafete, model.predict(suprafete_2d))

alte_puncte_prezise = [  [73], [80], [350]  ]
plt.scatter(alte_puncte_prezise, model.predict(alte_puncte_prezise))
```
