```python
from sklearn.datasets import make_regression
```

```python
x, y =  make_regression(n_samples=250, n_features=1)
plt.scatter(x, y)
plt.plot(x, y)
```

```python
x, y =  make_regression(n_samples=25, n_features=1, noise = 10)
plt.scatter(x, y)
# plt.plot(x, y)
```

```python
from scipy.stats import linregress

x, y =  make_regression(n_samples=25, n_features=1, noise = 10)
x = x.flatten()
lin_res = linregress(x, y)
plt.scatter(x, y)
plt.plot(x, lin_res.slope * x + lin_res.intercept);
```
