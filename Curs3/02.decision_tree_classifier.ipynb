```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Datasets/2_categories.csv")
df
```

```python
plt.scatter(df["coord_x"], df["coord_y"], c=df["category"]=="a");
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_model = DecisionTreeClassifier()
tree_model
```

```python
x = df[["coord_x", "coord_y"]] 
y = df["category"]


tree_model.fit(x, y)
```

# pip install graphviz

```python
from sklearn.tree import export_graphviz
import graphviz
```

```python
tree_graph = export_graphviz(tree_model, feature_names=["coord_x", "coord_y"],filled=True)
tree_graph
```

```python
graph = graphviz.Source(tree_graph)
graph
```
