```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model
```

```python
knn_model.n_neighbors
```

```python
x = df[["coord_x", "coord_y"]] 
y = df["category"] #.map({"a":1, "b":0})

knn_model.fit(x, y)
```

```python
knn_model.predict([[200, 220]])
```

```python
knn_model.predict([[500, 250]])
```

```python
knn_model.predict([[330, 350]])
```
