```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
fashion_test_path =  "fashion/fashion-mnist_test.csv"
fashion_train_path =  "fashion/fashion-mnist_train.csv"
```

```python
DATASET_LABELS = """0 T-shirt/top
1 Trouser
2 Pullover
3 Dress
4 Coat
5 Sandal
6 Shirt
7 Sneaker
8 Bag
9 Ankle boot
""".splitlines()
DATASET_LABELS = { int(item.split(" ", 1)[0]) : item.split(" ", 1)[1] for item in DATASET_LABELS}
DATASET_LABELS
```

```python
test_df = pd.read_csv(fashion_test_path)
```

```python
def show_image_at_index(index):
    if index not in range(0, 60000): return
    np_array_image = np.array(train_df.iloc[index][1:]).reshape(28, 28)
    image_label = DATASET_LABELS[train_df.iloc[index, 0]]
    plt.title(f"Image at index {index} - label {image_label}")
    plt.imshow(np_array_image.reshape(28, 28), cmap="gray");
```

# Citirea datelor din fisiere

```python
x_train = np.array(train_df.drop("label", axis=1))
x_train
```

```python
y_train = np.array(train_df["label"])
y_train
```

```python
x_train.shape, y_train.shape
```

```python
x_test = np.array(test_df.drop("label", axis=1))
y_test = np.array(test_df["label"])
```

```python
x_test.shape, y_test.shape
```

# Antrenarea modelului folosin scikit-learn

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = [RandomForestClassifier(), KNeighborsClassifier()] 
```

```python
scores = {}
```

```python
for model in models:
    model.fit(x_train, y_train)
    model_score = model.score(x_test, y_test)
    scores[model] = model_score
```
