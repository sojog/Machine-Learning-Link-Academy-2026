```python
def AND_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -1.5
    print(f"{x1} and {x2} = { 1 if  a1 * x1 + a2 * x2 + b > 0 else 0}") 


AND_neuron(1, 1)    # 1 + 1
AND_neuron(1, 0)    # 1
AND_neuron(0, 1)   # 1
AND_neuron(0, 0)
```

```python
def functie_de_activare(valoare):
    if valoare > 0:
        return 1
    else:
        return 0

def AND_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -1.5
    rezultat = functie_de_activare(a1 * x1 + a2 * x2 + b ) 
    print(f"{x1} and {x2} = {rezultat}") 


AND_neuron(1, 1)    # 1 + 1
AND_neuron(1, 0)    # 1
AND_neuron(0, 1)   # 1
AND_neuron(0, 0)
```

## Or clasic vs Neuroni

```python
def OR_classic(x1, x2):
    print(f"{x1} or {x2} = { x1 or x2 }") 


OR_classic(1, 1)    # 1 + 1
OR_classic(1, 0)    # 1
OR_classic(0, 1)   # 1
OR_classic(0, 0)
```

```python
def OR_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -0.5
    print(f"{x1} or {x2} = {  a1 * x1 + a2 * x2 + b}") 


OR_neuron(1, 1)    # 1 + 1
OR_neuron(1, 0)    # 1
OR_neuron(0, 1)   # 1
OR_neuron(0, 0)
```

```python
def functie_de_activare(valoare):
    if valoare > 0:
        return 1
    else:
        return 0
    

def OR_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -0.5
    print(f"{x1} or {x2} = {  functie_de_activare(a1 * x1 + a2 * x2 + b)}") 


OR_neuron(1, 1)    # 1 + 1
OR_neuron(1, 0)    # 1
OR_neuron(0, 1)   # 1
OR_neuron(0, 0)
```

# a1, a2 pot fi inlocuiti cu w1, w2

```python
def functie_de_activare(valoare):
    if valoare > 0:
        return 1
    else:
        return 0
    

def OR_neuron (x1, x2):
    w1 = 1
    w2 = 1
    b = -0.5
    print(f"{x1} or {x2} = {  functie_de_activare(w1 * x1 + w2 * x2 + b)}") 


OR_neuron(1, 1)    # 1 + 1
OR_neuron(1, 0)    # 1
OR_neuron(0, 1)   # 1
OR_neuron(0, 0)
```
