```python
def AND_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -1.5
    print(f"{x1} and {x2} = { 1 if  a1 * x1 + a2 * x2 + b > 0 else 0}") 


AND_neuron(1, 1)    # 1 + 1
AND_neuron(1, 0)    # 1
AND_neuron(0, 1)   # 1
AND_neuron(0, 0)
```

```python
def functie_de_activare(valoare):
    if valoare > 0:
        return 1
    else:
        return 0

def AND_neuron (x1, x2):
    a1 = 1
    a2 = 1
    b = -1.5
    rezultat = functie_de_activare(a1 * x1 + a2 * x2 + b ) 
    print(f"{x1} and {x2} = {rezultat}") 


AND_neuron(1, 1)    # 1 + 1
AND_neuron(1, 0)    # 1
AND_neuron(0, 1)   # 1
AND_neuron(0, 0)
```
