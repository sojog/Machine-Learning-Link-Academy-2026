# Folosind Tensorflow(Keras)

# wrapper

```python
tf.keras
```

```python
import keras
```

```python
keras
```

```python
model_secvential = tf.keras.models.Sequential()
model_secvential
```

```python
input_layer = tf.keras.layers.Input(shape=(1,))
input_layer
```

```python
model_secvential.add(input_layer)
model_secvential
```

```python
model_secvential.summary()
```

```python
first_dense_layer = tf.keras.layers.Dense(units=1, input_shape=(1,))
first_dense_layer
```

```python
model_secvential.add(first_dense_layer)
```

```python
model_secvential.summary()
```

```python
model_secvential
```

```python
model_secvential.compile(optimizer=tf.keras.optimizers.Adam(), loss=tf.keras.losses.mae)
```

```python
model_secvential.fit(x, y, epochs=100)
```

```python
model_secvential.predict(np.array([[10]]))
```
