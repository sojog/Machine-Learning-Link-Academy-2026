# Bag of words

```python
from sklearn.feature_extraction.text import CountVectorizer

# Mini dataset of sample reviews
# Creăm o mică listă cu 5 recenzii. Acesta este micul nostru „corpus” – setul de texte pe care îl analizăm.
corpus = [
    "I love this product",
    "This product is not good",
    "Absolutely fantastic experience",
    "Terrible, I hate it",
    "Not great, not terrible"
]

# Create and fit the vectorizer 
# Acesta este instrumentul BoW din scikit-learn care „urmărește” toate cuvintele și creează un dicționar (vocabular). El, automat:
# 1. găsește toate cuvintele diferite din corpus;
# 2. creează un dicționar (vocabular);
# 2. pentru fiecare recenzie creează un vector cu numere care arată de câte ori apare un cuvânt.
bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(corpus)

# Show feature names (vocabulary)
print("Vocabulary:", bow_vectorizer.get_feature_names_out())

# Convert sparse matrix to array for readability
print("\nBoW Matrix (Document-Term Matrix):\n", X_bow.toarray())
```
