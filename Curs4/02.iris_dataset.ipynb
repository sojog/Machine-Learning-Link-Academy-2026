```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
iris_ds = load_iris()
print(iris_ds.DESCR)
```

```python
x = pd.DataFrame(iris_ds.data, columns=iris_ds.feature_names)
x
```

```python
y = pd.DataFrame(iris_ds.target, columns=["categoria"])
y
```

```python
iris_ds.target_names
```

```python
plt.scatter(x["petal length (cm)"], x["petal width (cm)"], c=y["categoria"])
plt.xlabel("Petale")
```

```python
plt.scatter(x["sepal length (cm)"], x["sepal width (cm)"], c=y["categoria"])
plt.xlabel("Sepale")
```

```python
from sklearn.model_selection import train_test_split
```

```python
x.shape, y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
plt.scatter(x["petal length (cm)"], x["petal width (cm)"], c=y["categoria"])
plt.xlabel("Petale")
```

```python
plt.scatter(x_train["petal length (cm)"], x_train["petal width (cm)"], c=y_train["categoria"])

plt.xlabel("Petale - pentru antrenament")
```

```python
plt.scatter(x_test["petal length (cm)"], x_test["petal width (cm)"], c=y_test["categoria"])
```

# Date antrenament (rosu) vs date test (verde)

```python
plt.scatter(x_train["petal length (cm)"], x_train["petal width (cm)"], c="red")
plt.scatter(x_test["petal length (cm)"], x_test["petal width (cm)"], c="green")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
tree_model = DecisionTreeClassifier()
tree_model
```

```python
knn_model = KNeighborsClassifier()
knn_model 
```

```python
tree_model.fit(x_train, y_train)
```

```python
knn_model.fit(x_train, y_train)
```

```python
tree_model.score(x_test, y_test)
```

```python
knn_model.score(x_test, y_test)
```
