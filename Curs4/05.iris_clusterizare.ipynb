```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris_ds = load_iris()
print(iris_ds.DESCR)
```

```python
x = pd.DataFrame(iris_ds.data, columns=iris_ds.feature_names)
y = pd.DataFrame(iris_ds.target, columns=["categoria"])
```

```python
plt.scatter(x["petal length (cm)"], x["petal width (cm)"], c=y["categoria"])
plt.xlabel("Petale")
plt.legend(iris_ds.target_names)
```

```python
plt.scatter(x["sepal length (cm)"], x["sepal width (cm)"], c=y["categoria"])
plt.xlabel("Sepale")
plt.legend(iris_ds.target_names)
```

# Folosirea unui model de clusterizare (KMeans)

```python
from sklearn.cluster import KMeans
```

```python
kmeans_object = KMeans(n_clusters=3)
kmeans_object
```

```python
kmeans_object.fit(x)
```

```python
kmeans_object.predict(x)
```

```python
plt.scatter(x["petal length (cm)"], x["petal width (cm)"], c=kmeans_object.predict(x))
```

```python
plt.scatter(x["petal length (cm)"], x["petal width (cm)"], c=y["categoria"])
```
